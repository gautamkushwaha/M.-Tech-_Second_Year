**1. Insertion sort:**

Insertion sort is basically comparing the a element to its previous elements of the array , and if the element is smaller than the previous elements, the element is swapped until, it is smaller than those elements

````python


#not optimized for loop version
def insertion_sort1(A):
  for i in range(1,len(A)):
    for j in range (i-1, -1, -1):
      if A[j] > A[j+1]:
        A[j], A[j+1] = A[j+1], A[j]
      else:
        break

A = [5,9,1,2,4,8,6,3,7]
print(A)
insertion_sort1(A)
print(A)
````

````python
#optimized version
def insertion_sort3(A):
	for i in range(1, len(A)):
		curNum = A[i]
		k = 0
		for j in range(i-1, -2, -1):
			k = j
			if A[j] > curNum:
				A[j+1] = A[j]
			else:
				break
		A[k+1] = curNum


A = [5,9,1,2,4,8,6,3,7]
print(A)
insertion_sort3(A)
print(A)
````

**2. Selection sort**
first we assume the first element is the smallest element and search in the list for any element, larger than that. if found, we swap


````python
def selection_sort(A):
  for i in range(0, len(A)-1):
    min_index = i
    for j in range(i+1, len(A)):
      if A[j] < A[min_index]:
        min_index = j
    if min_index != i:
        A[i], A[min_index] = A[min_index], A[i]

A = [5,9,1,2,4,8,6,3,7]
print(A)
selection_sort(A)
print(A)
````

5. Quick sort

````python
# partition function
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] < pivot:
            i += 1
            swap(arr, i, j)
    swap(arr, i + 1, high)
    return i + 1

def swap(arr, i, j):
    arr[i], arr[j] = arr[j], arr[i]

def quickSort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)
        quickSort(arr, low, pi - 1)
        quickSort(arr, pi + 1, high)

if __name__ == "__main__":
    arr = [10, 7, 8, 9, 1, 5]
    n = len(arr)
    quickSort(arr, 0, n - 1)
    for val in arr:
        print(val, end=" ")
````

# Consine similarity of two vectors
It measures the cosine of the angle between two non-zero vectors in an inner product space.

If the two vectors point in the exact same direction, the similarity is 1. If they are perpendicular it is 0.

image.png

````python
#python implementation(using Numpy)

import numpy as np

def cal_cos_sim(vec1, vec2):
  # convert list to numpy array
  a = np.array(vec1)
  b = np.array(vec2)

  # calculat the dot product
  dot_product = np.dot(a,b)

  # calculate the magnitute  (norm) of each vector

  norm_a = np.linalg.norm(a)
  norm_b = np.linalg.norm(b)

  # calculate the cosine similarity
  simi = dot_product / (norm_a * norm_b)

  return simi

# example vector

vector_a = [1, 2, 3]
vector_b = [ 1, 2, 3]

result = cal_cos_sim(vector_a, vector_b)

````

````python
from collections import deque

def count_students_unable_to_eat(students, sandwiches):
    # Convert lists to deques for efficient popleft operations
    students_queue = deque(students)
    sandwiches_stack = deque(sandwiches)

    # Counter for consecutive rejections to detect deadlock
    # If a student doesn't get their preferred sandwich, they go to the end of the queue.
    # If the same student keeps coming back to the front and not getting their sandwich
    # this means no one else can get their sandwich either.
    rejection_count = 0

    while students_queue and sandwiches_stack and rejection_count < len(students_queue):
        student_preference = students_queue[0]
        current_sandwich = sandwiches_stack[0]

        if student_preference == current_sandwich:
            # Student takes the sandwich
            students_queue.popleft()
            sandwiches_stack.popleft()
            rejection_count = 0  # Reset rejection count as a sandwich was taken
        else:
            # Student doesn't like the sandwich, goes to the end of the queue
            students_queue.append(students_queue.popleft())
            rejection_count += 1 # Increment rejection count

    # The number of students unable to eat is the remaining students in the queue
    return len(students_queue)

# Example usage:
# students = [1,1,0,0] (1 for circular sandwich, 0 for square sandwich)
# sandwiches = [0,1,0,1]

students_example1 = [1,1,0,0]
sandwiches_example1 = [0,1,0,1]
result1 = count_students_unable_to_eat(students_example1, sandwiches_example1)
print(f"Example 1: Students: {students_example1}, Sandwiches: {sandwiches_example1}")
print(f"Number of students unable to eat: {result1}") # Expected output: 0

students_example2 = [1,1,1,0,0,1]
sandwiches_example2 = [1,0,0,0,1,1]
result2 = count_students_unable_to_eat(students_example2, sandwiches_example2)
print(f"Example 2: Students: {students_example2}, Sandwiches: {sandwiches_example2}")
print(f"Number of students unable to eat: {result2}") # Expected output: 3 (students [0,0,1] remain)

students_example3 = [0,0,0]
sandwiches_example3 = [1,1,1]
result3 = count_students_unable_to_eat(students_example3, sandwiches_example3)
print(f"Example 3: Students: {students_example3}, Sandwiches: {sandwiches_example3}")
print(f"Number of students unable to eat: {result3}") # Expected output: 3
````

````python
# step 1 load pre-trained bert model and tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')
````

````python
# step 2: prepare sentence with same word in different contexts

sentences = {
    " I deposited money in the bank",
    " WE sat by the river bank"
}
````

````python
# step 3: Get contextual Embeddings

def get_contexual_embeddings(sentence):
  # Tokenize
  inputs = tokenizer(sentence, return_tensors='pt', padding= True, truncation= True)

  # Get embeddings
  with torch.no_grad():
    outputs = model(**inputs)

  # Last hidden states(contextual Embeddings)
  embeddings = outputs.last_hidden_state

  return embeddings

# step 4 compare embeddings for bank in different context
for sentence in sentences:
  embeddings = get_contexual_embeddings(sentence)
  tokens = tokenizer.tokenize(sentence)

  print(f"\nSentence: {sentence}")
  print(f"\n Tokens:{tokens}")
  print(f"\nEmbeddings: {embeddings}")


  # find position of bank

  if 'bank' in sentence:
    bank_idx = tokens.index('bank')
    bank_embedding = embeddings[0, bank_idx : 10] # first 10 dimensions
    print(f"\n Bankk embedding (first 10 dime): {bank_embedding.numpy()}")
````

````python
# Function to get embedding for a single word
def get_word_embedding(word, tokenizer, model):
    # Tokenize the word directly (or a sentence containing it)
    # For a single word, we can wrap it in a sentence for context
    sentence = f"This is a {word}."
    inputs = tokenizer(sentence, return_tensors='pt', padding=True, truncation=True)

    with torch.no_grad():
        outputs = model(**inputs)

    # Get the last hidden state (contextual embeddings)
    embeddings = outputs.last_hidden_state

    # Find the index of the target word
    tokens = tokenizer.tokenize(sentence)
    try:
        # Find the actual token corresponding to the word
        word_token_idx = -1
        for i, token in enumerate(tokens):
            if word.lower() in token.lower(): # Case-insensitive check, includes subwords
                word_token_idx = i
                break

        if word_token_idx != -1:
            # Return the embedding for the specific word token
            return embeddings[0, word_token_idx, :]
        else:
            print(f"Could not find token for '{word}' in: {tokens}")
            return None
    except ValueError:
        print(f"Word '{word}' not found in the tokenized sentence.")
        return None

# Example: Embed a single word
word_to_embed = "king"
king_embedding = get_word_embedding(word_to_embed, tokenizer, model)

if king_embedding is not None:
    print(f"\nEmbedding for '{word_to_embed}' (first 10 dimensions):\n{king_embedding.numpy()[:10]}...")
    print(f"\nEmbedding shape: {king_embedding.shape}")
````

### How Embedding is Done in BERT

BERT (Bidirectional Encoder Representations from Transformers) generates embeddings in a sophisticated, contextual manner. Unlike traditional word embeddings (like Word2Vec or GloVe) that assign a fixed vector to each word, BERT's embeddings change based on the word's context in a sentence.

Here's a breakdown of the process:

1.  **Tokenization**: The input text is first broken down into a sequence of tokens using a WordPiece tokenizer. This tokenizer handles subword units, allowing it to represent out-of-vocabulary words. Special tokens like `[CLS]` (for classification tasks) and `[SEP]` (to separate sentences) are added.

2.  **Input Embedding Layer**: For each token, BERT combines three types of embeddings:
    *   **Token Embeddings**: These are the standard word embeddings, representing the semantic meaning of the token itself, learned during pre-training.
    *   **Segment Embeddings**: These indicate which sentence a token belongs to (e.g., `sentence A` or `sentence B`). This is important for tasks like Question Answering where two segments are compared.
    *   **Position Embeddings**: These capture the positional information of each token within the sequence, as Transformers are permutation-invariant and need to know the order of words.

    These three embeddings are summed up to create a single input embedding vector for each token.

3.  **Transformer Encoder Layers**: These combined input embeddings are then fed through multiple layers of a Transformer encoder (12 layers for `bert-base`, 24 for `bert-large`). Each Transformer layer consists of:
    *   **Multi-Head Self-Attention**: This mechanism allows each token to weigh the importance of all other tokens in the sequence when generating its own representation. It's "bidirectional" because it considers context from both left and right.
    *   **Feed-Forward Network**: A standard neural network applied independently to each position.

    The self-attention mechanism is key to generating *contextual* embeddings. For instance, the word "bank" will have different attention patterns and thus different final representations depending on whether it's used in the context of a "river bank" or a "money bank."

4.  **Output Embeddings**: The output of the final Transformer encoder layer for each token is its contextual embedding. These vectors capture rich semantic and syntactic information, influenced by the entire input sequence.


### Are Embeddings Done the Same Way Across All Models?

No, the way embeddings are generated varies significantly across different models, especially between traditional and contextual models:

*   **Traditional (Non-Contextual) Models (e.g., Word2Vec, GloVe, FastText)**:
    *   Assign a *fixed, static* vector to each word in the vocabulary. The embedding for "bank" is always the same, regardless of its usage in a sentence. They capture semantic relationships between words based on co-occurrence in a corpus.
    *   Mathematically, they learn a mapping from a one-hot encoded word to a dense vector space.

*   **Contextual Models (e.g., BERT, GPT, RoBERTa, XLNet)**:
    *   Generate *dynamic* embeddings where the vector representation of a word changes depending on the surrounding words (its context) in a given sentence. This allows them to handle polysemy (words with multiple meanings).
    *   These models typically use Transformer architectures with self-attention mechanisms to incorporate context.
    *   The core idea is to process the entire input sequence simultaneously, allowing each word's representation to be influenced by all other words.

*   **Hybrid Approaches**: Some models might use a combination. For instance, older models might have a static embedding layer followed by a recurrent neural network that adds some context, but it's not as deep or bidirectional as a Transformer.

In summary, while the goal of generating numerical representations of text is common, the underlying architectures (e.g., shallow neural networks vs. deep Transformers) and the resulting properties of the embeddings (static vs. dynamic, contextual vs. non-contextual) differ greatly.

### Mathematical Simplification of Embedding

At its simplest, an embedding can be thought of as a mapping function that transforms a discrete item (like a word from a vocabulary) into a continuous vector space. While complex in neural networks, a basic analogy can be described mathematically.

Let's consider a vocabulary $V$ of size $N$ words: $V = \{w_1, w_2, \dots, w_N\}$.

For any word $w_i \in V$, its embedding is a vector $e_i \in \mathbb{R}^d$, where $d$ is the dimensionality of the embedding space (e.g., 768 for BERT).

**1. Simplest Non-Contextual Embedding (e.g., Word2Vec conceptualization):**

Imagine a lookup table, $M$, where each row corresponds to a word in your vocabulary. If you represent each word as a one-hot vector (a vector with a 1 at the index corresponding to the word and 0s elsewhere), then getting an embedding is essentially a matrix multiplication.

Let $w_{\text{one-hot}}$ be the one-hot vector for a word (e.g., `[0, 0, ..., 1, ..., 0]`).
Let $W_{\text{embed}}$ be an $N \times d$ matrix, where $N$ is the vocabulary size and $d$ is the embedding dimension.

The embedding $e$ for that word is:

$$e = w_{\text{one-hot}} \times W_{\text{embed}}$$

This simply selects the row from $W_{\text{embed}}$ corresponding to the word. The values in $W_{\text{embed}}$ are learned during training.

**2. Conceptualizing Contextual Embeddings (Simplified):**

For a contextual model like BERT, the process is far more complex because the embedding of a word also depends on the *entire input sequence*. It's not just a lookup. Instead, it involves multiple transformations through neural network layers.

Let $S = (w_1, w_2, \dots, w_k)$ be an input sentence of $k$ words.

For each word $w_i$ in the sentence $S$, its contextual embedding $e'_i$ is a function of the entire sentence:

$$e'_i = f(w_i, S) = f(w_i, w_1, w_2, \dots, w_k)$$

In a Transformer model, $f$ involves:
*   Initial token, segment, and position embeddings (which are sums of vectors).
*   Multiple layers of self-attention and feed-forward networks.

Each self-attention head computes a weighted sum of transformed token representations. For a token $i$, its output depends on its query vector ($Q_i$), and the key ($K_j$) and value ($V_j$) vectors of all other tokens $j$ in the sequence:

$$ \text{Attention}(Q, K, V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V $$

Where:
*   $Q$ are query vectors, $K$ are key vectors, $V$ are value vectors (derived from initial token embeddings).
*   $d_k$ is the dimension of the key vectors.

This output is then passed through a feed-forward network and other layers. The contextual embedding for $w_i$ is the final output vector for that token after all Transformer layers. The key is that $QK^T$ involves dot products between $Q_i$ and all $K_j$, meaning token $i$ 'attends' to all other tokens to form its new representation.

Example 2 : Simple Comparison Script

````python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from transformers import BertTokenizer, BertModel # Import necessary classes

# Ensure the tokenizer and model are Bert-specific for this function
tokenizer_bert = BertTokenizer.from_pretrained('bert-base-uncased')
model_bert = BertModel.from_pretrained('bert-base-uncased')

def get_contexual_embeddings_bert(sentence):
  # Tokenize using the BERT tokenizer
  inputs = tokenizer_bert(sentence, return_tensors='pt', padding= True, truncation= True)

  # Get embeddings using the BERT model
  with torch.no_grad():
    outputs = model_bert(**inputs)

  # Last hidden states(contextual Embeddings)
  embeddings = outputs.last_hidden_state

  return embeddings

def compare_contexts():
  sentences = [
      "The bank charges interest on loans",
      "The river bank was muddy after rain",
      "He works at a bank downtown"
  ]

  all_embeddings = []


  for sentence in sentences:
    # Use the BERT-specific contextual embeddings function
    embeddings = get_contexual_embeddings_bert(sentence)
    tokens = tokenizer_bert.tokenize(sentence)

    # Find the position of 'bank' in the tokens for the current sentence
    # Note: BERT tokenizer might split 'bank' into subwords, but for simplicity, we look for 'bank' directly
    bank_token_idx = -1
    for i, tok in enumerate(tokens):
        if 'bank' in tok.lower(): # Use 'tok' from this loop, and handle subwords/case
            bank_token_idx = i
            break

    if bank_token_idx != -1:
        # Extract the embedding for the 'bank' token from the first (and only) sentence in the batch
        bank_embedding = embeddings[0, bank_token_idx, :]
        all_embeddings.append(bank_embedding)
    else:
        print(f"Warning: 'bank' token not found in tokenized sentence: {tokens}")
        # Append a placeholder or handle this case as appropriate, e.g., skip this sentence
        all_embeddings.append(None) # Appending None for now to maintain list length if word not found

  # Filter out None values if 'bank' was not found in some sentences
  all_embeddings = [emb for emb in all_embeddings if emb is not None]

  # Calculate similarities only if there are at least two embeddings
  if len(all_embeddings) < 2:
      print("Not enough 'bank' embeddings to compare similarities.")
      return

  print("\nCosine Similarities between 'bank' in different contexts:")
  for i in range(len(all_embeddings)):
    for j in range(i+1, len(all_embeddings)):
      sim = cosine_similarity(
          all_embeddings[i].reshape(1, -1),
          all_embeddings[j].reshape(1, -1)
      )[0][0]
      print(f" '{sentences[i][:20]}....' vs '{sentences[j][:20]}....': {sim:.4f}")


# Run Comparison
compare_contexts()
````

# **Batch Size & BERT Embeddings Shape Explained**

## **1. What is Batch Size?**

### **Analogy:**
Think of batch processing like cooking:
- **Single dish** (batch_size=1): Cook one meal at a time
- **Multiple dishes** (batch_size=32): Cook 32 meals simultaneously

### **Technical Definition:**
- **Batch Size** = Number of samples processed together in one forward pass
- Instead of processing sentences one-by-one, process multiple sentences together for efficiency

---

## **2. BERT Embeddings Shape: [batch_size, sequence_length, hidden_size]**

Let's break down each dimension:

### **Dimension 1: `batch_size`**
```python
# Example: Processing 4 sentences together
sentences = [
    "I love programming",
    "She likes reading",
    "He enjoys sports",
    "They watch movies"
]

# Batch size = 4 (all processed together)
embeddings.shape[0] == 4
```

### **Dimension 2: `sequence_length`**
```python
# For sentence: "I love programming"
tokens = ['[CLS]', 'i', 'love', 'programming', '[SEP]']
# sequence_length = 5 (including special tokens)
```

**Special Tokens BERT adds:**
- `[CLS]`: Classification token (first position)
- `[SEP]`: Separator token (end of sentence)
- Padding tokens if needed

**Example with padding:**
```python
sentence1: "Hello" → tokens: ['[CLS]', 'hello', '[SEP]', '[PAD]', '[PAD]']  # length=5
sentence2: "Good morning world" → tokens: ['[CLS]', 'good', 'morning', 'world', '[SEP]']  # length=5

# After padding, both have sequence_length = 5
```

### **Dimension 3: `hidden_size`**
```python
# Different BERT models have different hidden sizes:
# bert-base-uncased: 768 dimensions
# bert-large-uncased: 1024 dimensions
# distilbert-base-uncased: 768 dimensions
```

**What's in these 768 numbers?**
Each dimension captures different linguistic features:
- Dimensions 0-100: Might capture "verb-ness"
- Dimensions 101-200: Might capture "noun-ness"
- Dimensions 201-300: Might capture "plural form"
- Dimensions 301-400: Might capture "past tense"
- etc.

- **Large batch size (64+)**: More memory, faster training, but may affect quality

## **6. Memory Calculation**

```python
# Calculate memory usage
def calculate_memory(batch_size, seq_len, hidden_size=768, dtype=torch.float32):
    # Each float32 = 4 bytes
    tensor_size = batch_size * seq_len * hidden_size
    memory_bytes = tensor_size * 4
    memory_mb = memory_bytes / (1024 * 1024)
    
    return memory_mb

# Example calculations
print(f"Batch 1, length 128: {calculate_memory(1, 128):.2f} MB")
print(f"Batch 32, length 128: {calculate_memory(32, 128):.2f} MB")
print(f"Batch 64, length 512: {calculate_memory(64, 512):.2f} MB")
```

**Key Points to Remember:**
1. **batch_size**: How many sentences processed together
2. **sequence_length**: Max tokens in the batch (with padding)
3. **hidden_size**: Dimension of each token's embedding
4. The shape `[batch_size, sequence_length, hidden_size]` allows BERT to process multiple sentences efficiently while maintaining context for each token!
